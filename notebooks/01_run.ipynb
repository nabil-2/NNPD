# Train and analyze
Training choices live in `settings_training.py`, analysis choices in `settings_analysis.py`. This notebook runs the small `smoke` profile; `python run.py run --profile smoke` does the same. A `Choice` in the training settings makes a baseline-plus-one-factor sweep, not a Cartesian product.

```python
from pathlib import Path
import os
import sys

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / "settings_training.py").is_file() and (p / "nnpd").is_dir())
os.chdir(ROOT)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from nnpd import execute, load_settings, plan
from nnpd.results import runs

PROFILE = "smoke"  # or "default" / "paper"; use the same profile in all three notebooks
OUTPUT = load_settings("settings_training.py", PROFILE)["output"]
```

```python
jobs = plan(PROFILE)
[(job.changed or "baseline", job.member, job.workload) for job in jobs]
```

## Execute
`run` trains the models, then analyzes them. The four priors form a fixed comparison cohort. A repeat reuses the trained models and cached analysis data; interrupted training restarts its uncommitted stage.

```python
paths = execute("run", profile=PROFILE)
paths
```

```python
[(record["member"]["prior"], record["metrics"]["classification"]) for record in runs(OUTPUT)]
```
